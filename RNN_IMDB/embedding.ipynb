```python
from tensorflow import keras
from tensorflow.keras.preprocessing.text import one_hot
```

```python
## sentences
sent=['the glass of milk',
      'the glass of juice',
      'the cup of tea',
      'I am a good boy',
      'I am a good developer',
      'under the tree',
      'the bird is flying in the sky',
      'the lion is the king of the jungle'
      ]
sent
```

```python
## Define the vocabulary size
voc_size=10000
```

```python
## One hot representation of the sentences
one_hot_rep=[one_hot(words,voc_size)for words in sent]
one_hot_rep
```

What happend is all the given sentences is converted into a numerical vector representation also caaled as feature vector

```python
# Word Embedding Representation

from tensorflow.keras.layers import Embedding
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
```

```python
import numpy as np
```

```python
sent_length=8
embedded_docs=pad_sequences(one_hot_rep,padding='pre',maxlen=sent_length)
print(embedded_docs)
```

```python
dim=10

model = Sequential()

model.add(
    Embedding(
        input_dim=voc_size,
        output_dim=dim,
        input_shape=(sent_length,)
    )
)
```

```python
model.summary()
```

```python
model.predict(embedded_docs)

#  every word in the vocabulary will be represented as a vector of size 10. 
# The embedding layer will learn these representations during training. 
# The input to the embedding layer is a sequence of integers (the 
# one-hot encoded representation of the sentences), and the output is a 
# sequence of dense vectors (the learned embeddings).
```

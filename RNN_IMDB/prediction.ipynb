```python
import numpy as np
import re
import tensorflow as tf
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing import sequence
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding,SimpleRNN,Dense
```

```python
# Mapping of word indices to words(for understanding the review)

word_index=imdb.get_word_index()
word_index
reverse_word_index={value:key for key,value in word_index.items()}

reverse_word_index
```

```python
# load the pre-trained model with ReLU activation 
# Load the pre-trained model with ReLU activation
model = tf.keras.models.load_model("simple_rnn_imdb.h5")

model.summary()
```

```python
# Model Weights
model.get_weights()
```

```python
# Step 2: Helper Functions
# Function to decode reviews
def decode_review(encoded_review):
    return ' '.join([reverse_word_index.get(i - 3, '?') for i in encoded_review])

# Function to preprocess user input
import re

def preprocess_text(text):
    text = re.sub(r'[^a-zA-Z0-9\s]', '', text.lower())
    words = text.split()

    encoded_review = [
        word_index[word] + 3 if word in word_index else 2
        for word in words
    ]

    padded_review = sequence.pad_sequences(
        [encoded_review],
        maxlen=500
    )

    return padded_review
```

```python
### Prediction  function

def predict_sentiment(review):
    preprocessed_input=preprocess_text(review) # preprocess the input

    prediction=model.predict(preprocessed_input)

    sentiment = 'Positive' if prediction[0][0] > 0.5 else 'Negative'
    
    return sentiment, prediction[0][0]

```

```python
# Step 4: User Input and Prediction

# Example review for prediction
max_len=500  # Define the maximum length for padding
example_review = "This movie was very fantastic."

sentiment, score = predict_sentiment(example_review)

print(f'Review: {example_review}')
print(f'Sentiment: {sentiment}, Score: {score}')
```

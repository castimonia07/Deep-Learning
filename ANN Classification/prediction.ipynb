```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
import pickle
import pandas as pd
import numpy as np
```

```python
## Load the Trained Model,scalar pickle,one hot encoder pickle 

model = tf.keras.models.load_model('model.h5')

## laod the encoder and scaler 
with open('onehot_encoder_geo.pkl','rb') as file:
    label_encoder_geo=pickle.load(file)  # Load the label encoder for gender from the pickle

with open("label_encoder_gender.pkl", "rb") as file:
    label_encoder_gender = pickle.load(file)  # Load the label encoder for gender from the

with open('scaler.pkl', 'rb') as file:
    scaler = pickle.load(file)  # Load the scaler from the pickle file
```

```python
#Example input data

# Example input data
input_data = {
    'CreditScore': 600,
    'Geography': 'France',
    'Gender': 'Male',
    'Age': 40,
    'Tenure': 3,
    'Balance': 60000,
    'NumOfProducts': 2,
    'HasCrCard': 1,
    'IsActiveMember': 1,
    'EstimatedSalary': 50000
}
```

```python
# One-hot encode 'Geography'
geo_encoded = label_encoder_geo.transform([[input_data['Geography']]]).toarray()
geo_encoded_df = pd.DataFrame(geo_encoded, columns=label_encoder_geo.get_feature_names_out(['Geography']))
geo_encoded_df
```

```python
input_df=pd.DataFrame([input_data])
input_df
```

```python
## Encode categorical variables
input_df['Gender']=label_encoder_gender.transform(input_df['Gender'])
input_df
```

```python
#Combine one-hot encoded 'Geography' with the rest of the input data
# Combine one-hot encoded Geography with the remaining input data
input_df = pd.concat(
    [input_df.drop(columns=["Geography"]), geo_encoded_df],
    axis=1
)

input_df
```

```python
## Scaling the input data using the loaded scaler

scaled_input_df = scaler.transform(input_df)
scaled_input_df
```

```python
## Predict the output using the trained model
prediction=model.predict(scaled_input_df)
prediction
```

```python
# prediction probability
prediction_prob=prediction[0][0]
```

```python
prediction_prob
```

```python
if prediction_prob>0.5:
    print("The customer is likely to churn.")
else:
    print("The customer is not likely to churn.")
```
